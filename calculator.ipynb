```python
##### print("===== MULTI-NUMBER CALCULATOR =====")

while True:
    print("\nChoose an operation:")
    print("+. Addition (+)")
    print("-. Subtraction (-)")
    print("*. Multiplication (*)")
    print("/. Division (/)")
    print("exit.to close calculator")

    choice = input("Enter your choice: ")

    if choice == "exit":
        confirm=input("are you want to exit (yes/no)")
        if confirm.lower()=="yes":
            print("Thankyou for using calculator")
            break
        else:
            continue
    if choice not in ["+", "-", "*", "/"]:
        print("Invalid choice!")
        print("USE RIGTH OPERATOR TO USE THE CALCULATOR")
        break
    numbers = input(
        "Enter numbers separated by spaces: "
    ).split()

    try:
        numbers = [float(n) for n in numbers]

        if len(numbers) == 0:
            print("Please enter at least one number.")
            continue

        if choice == "+":
            result = sum(numbers)

        elif choice == "-":
            result = numbers[0]
            for n in numbers[1:]:
                result -= n

        elif choice == "*":
            result = 1
            for n in numbers:
                result *= n

        elif choice == "/":
            result = numbers[0]
            for n in numbers[1:]:
                if n == 0:
                    print("Cannot divide by zero!")
                    break
                result /= n
            else:
                print("Result:", result)
                continue
            continue

        print("Result:", result)

    except ValueError:
        print("Please enter valid numbers!")
```
